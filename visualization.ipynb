```python
#loading
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## pie chart

```python
#Plot a pie chart showing the percentage distribution of Gender
gender_counts = df['Gender'].value_counts()

# Plot pie chart
plt.figure()
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%')
plt.title("Percentage Distribution of Gender")
plt.show()
```

## Barchart

```python
#Create a bar chart showing the number of customers in each City.
city_counts = df['City'].value_counts()

# Plot bar chart
plt.figure()
plt.bar(city_counts.index, city_counts.values)
plt.title("Number of Customers in Each City")
plt.xlabel("City")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.show()
```

```python
#. Plot a line chart to visualize Purchase Amount trend over Purchase Date.
import pandas as pd

df = pd.read_csv("Practice-Datasets .csv", sep=",")
print(df.columns)
# Convert purchase_date to datetime
df['Purchase_Date'] = pd.to_datetime(df['Purchase_Date'])

# Sort by date
df = df.sort_values('Purchase_Date')

# Group by date
daily_sales = df.groupby('Purchase_Date')['Purchase_Amount'].sum()

# Plot line chart
plt.figure()
plt.plot(daily_sales.index, daily_sales.values)
plt.title("Purchase Amount Trend Over Time")
plt.xlabel("Purchase Date")
plt.ylabel("Total Purchase Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
#. Create a bar plot to compare average Monthly Income by Gender.
import matplotlib.pyplot as plt

plt.figure()
avg_income = df.groupby('Gender')['Monthly_Income'].mean()
print(avg_income)
plt.bar(avg_income.index, avg_income.values)
plt.title("Average Monthly Income by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Monthly Income")
plt.show()
```

```python
# Plot a box plot of Purchase Amount grouped by Product Category.
plt.figure()

# Get unique categories
categories = df['Product_Category'].unique()

# Prepare data grouped by category
data = [df[df['Product_Category'] == cat]['Purchase_Amount'] for cat in categories]

# Create box plot
plt.boxplot(data)
plt.xticks(range(1, len(categories)+1), categories, rotation=45)
plt.title("Purchase Amount by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Purchase Amount")
plt.show()
```

```python
#Create a pie chart to show Product Category distribution
plt.figure()
category_counts = df['Product_Category'].value_counts()
print(category_counts)
plt.pie(category_counts.values,
        labels=category_counts.index,
        autopct='%1.1f%%')
plt.title("Product Category Distribution")
plt.show()
```

```python
# Plot a histogram of Purchase Amount with appropriate bin size.
plt.figure()
plt.hist(df['Purchase_Amount'], bins='auto')
plt.title("Histogram of Purchase Amount")
plt.xlabel("Purchase Amount")
plt.ylabel("Frequency")
plt.show()
```

```python
# Create a bar chart showing average Spending Score by City.
plt.figure()
avg_spending = df.groupby('City')['Spending_Score'].mean()
print(avg_spending)
plt.bar(avg_spending.index, avg_spending.values)
plt.title("Average Spending Score by City")
plt.xlabel("City")
plt.ylabel("Average Spending Score")
plt.xticks(rotation=45)
plt.show()
```

```python
#Plot a line chart showing total Purchase Amount per month
plt.figure()
df['year_month'] = df['Purchase_Date'].dt.to_period('M')
monthly_sales = df.groupby('year_month')['Purchase_Amount'].sum()

plt.plot(monthly_sales.index.astype(str), monthly_sales.values)
plt.title("Total Purchase Amount per Month")
plt.xlabel("Month")
plt.ylabel("Total Purchase Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
#Create a box plot comparing Monthly Income across different Cities.
plt.figure()

# Get unique cities
cities = df['City'].unique()

# Prepare income data grouped by city
data = [df[df['City'] == city]['Monthly_Income'] for city in cities]

# Create box plot
plt.boxplot(data)
plt.xticks(range(1, len(cities)+1), cities, rotation=45)
plt.title("Monthly Income Distribution by City")
plt.xlabel("City")
plt.ylabel("Monthly Income")
plt.show()
```

```python
#Plot a histogram to analyze Age distribution by Gender.
plt.figure()

# Separate data by gender
male_age = df[df['Gender'] == 'Male']['Age']
female_age = df[df['Gender'] == 'Female']['Age']

# Plot histograms
plt.hist(male_age, bins=10, alpha=0.5, label='Male')
plt.hist(female_age, bins=10, alpha=0.5, label='Female')

plt.title("Age Distribution by Gender")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
#. Create a bar chart showing total Purchase Amount by Product Category.
plt.figure()
total_sales = df.groupby('Product_Category')['Purchase_Amount'].sum()
plt.bar(total_sales.index, total_sales.values)
plt.title("Total Purchase Amount by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Purchase Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
#. Plot a scatter plot between Monthly Income and Spending Score.
plt.figure()
plt.scatter(df['Monthly_Income'], df['Spending_Score'])
plt.title("Monthly Income vs Spending Score")
plt.xlabel("Monthly Income")
plt.ylabel("Spending Score")
plt.show()
```

```python
#Create a line chart to visualize daily Purchase Amount variation.
plt.figure()
daily_sales = df.groupby('Purchase_Date')['Purchase_Amount'].sum()
print(daily_sales.head())
plt.plot(daily_sales.index, daily_sales.values)
plt.title("Daily Purchase Amount Variation")
plt.xlabel("Date")
plt.ylabel("Total Purchase Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
# Plot a box plot of Spending Score grouped by Gender
df.boxplot(column='Spending_Score', by='Gender')
plt.title("Spending Score by Gender")
plt.suptitle("")   # removes extra default title
plt.xlabel("Gender")
plt.ylabel("Spending Score")
plt.show()
```

```python
#Create a bar chart showing customer count by Product Category and Gender
category_gender_counts = pd.crosstab(df['Product_Category'], df['Gender'])
category_gender_counts.plot(kind='bar')
plt.title("Customer Count by Product Category and Gender")
plt.xlabel("Product Category")
plt.ylabel("Customer Count")
plt.xticks(rotation=45)
plt.legend(title="Gender")
plt.show()
```

```python
#Analyze the relationship between Age and Purchase Amount using a suitable visualization.
plt.figure()
plt.scatter(df['Age'], df['Purchase_Amount'])
plt.title("Age vs Purchase Amount")
plt.xlabel("Age")
plt.ylabel("Purchase Amount")
plt.show()
```

```python
#Plot a box plot to analyze Monthly Income distribution
plt.figure()
plt.boxplot(df['Monthly_Income'].dropna())
plt.title("Box Plot of Monthly Income Distribution")
plt.ylabel("Monthly Income")
plt.show()
```

```python
#Create a histogram of Spending Score and interpret the skewness
plt.figure()
plt.hist(df['Spending_Score'], bins=15)
plt.title("Histogram of Spending Score")
plt.xlabel("Spending Score")
plt.ylabel("Frequency")
plt.show()
```
